## 1. Explore one group/dataset at a time (including its subgroups/datasets)


### Select data to access and enter one of the following options for data_primary variable.
### data_primary = "Dpu", "LandscapeScenario", "PPM", "Pearl", "Swap", or "Weather"

```python
data_primary = "Pearl"
```

```python

# Open the HDF5 file
with h5py.File(arrdat_path, 'r') as file:
    # Access the dataset group by name (replace data_primary variable to change this)
    dataset = file[data_primary]
    
    # Check if the dataset is a group or a dataset
    if isinstance(dataset, h5py.Group):
        # If group, access the members
        for name in dataset:
            subgroup = dataset[name]
            print(subgroup.name)  
            # Access subgroup's data or further subgroups if needed
    elif isinstance(dataset, h5py.Dataset):
        # If dataset, access directly
        data = dataset[:]
        print(data)
```

### Now you can select a dataset to learn more about by choosing one of the names printed above to define the variable "data_secondary".
### Then access and print "data_secondary", the data type, as well as the dimensions

```python
data_secondary = "AmaUptPro"
```

```python
with h5py.File(arrdat_path, 'r') as file:
    
    group = file[data_primary]
    
    # Access a specific subgroup by its name (printed above, now defined as "data_secondary")
    subgroup_name = data_secondary  
    subgroup = group[subgroup_name]
    
    #access data or further subgroups within this subgroup
    if isinstance(subgroup, h5py.Dataset):
        data = subgroup[:]
        # Use the data as needed
        print(data)
        print(type(data))
        dimensions = data.shape
        print("Dimensions:", dimensions)
```

### View all of the groups and datasets by running the next chunk (see everything but the actual data values in the hdf5)
### Make sure to click "scrollable element" at the bottom of the printed result to avoid truncation

```python
def print_hdf5_tree(hdf5_group, parent_group=''):
    for key in hdf5_group.keys():
        if isinstance(hdf5_group[key], h5py.Group):
            print(f"Group: {parent_group}/{key}")
            print_hdf5_tree(hdf5_group[key], f"{parent_group}/{key}")
        else:
            print(f"Dataset: {parent_group}/{key}")

with h5py.File(arrdat_path, 'r') as hdf5_file:
    print("HDF5 Tree Structure:")
    print_hdf5_tree(hdf5_file)
```
